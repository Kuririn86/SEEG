```python
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'pyproject.toml').exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / 'src'))
DATA_ROOT = PROJECT_ROOT / 'Dataset' / 'sampleData'
```

## Data

```python
from seeg_detector.analysis import build_profile
profile = build_profile(DATA_ROOT)
profile.summary
```

## Results

```python
profile.files[['sample_id', 'split', 'label', 'channels', 'duration_seconds', 'prefix_bytes']]
```

```python
from seeg_detector.data import load_record
from seeg_detector.visualization.waveforms import plot_onset_comparison
labels = profile.labels.set_index('sample_id')
positive_id, negative_id = 'train_000027', 'train_000057'
positive_row = labels.loc[positive_id]
channels = [str(positive_row[f'ch{i}']) for i in range(1, 7)]
fig = plot_onset_comparison(
    load_record(DATA_ROOT / f'{positive_id}.npz'),
    load_record(DATA_ROOT / f'{negative_id}.npz'),
    float(positive_row['onset_time']),
    channels,
)
fig
```
